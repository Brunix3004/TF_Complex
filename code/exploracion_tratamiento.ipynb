```python
import pandas as pd
```

```python
data = pd.read_csv("./../datasets/datosRed.csv", encoding="ISO-8859-1")
data.head()
```

```python
data.shape
```

```python
data.info()
```

NO HAY FALTANTES

```python
#Total duplicados
data.duplicated().sum()
```

```python
#Mostrar filas duplicadas
duplicados = data[data.duplicated(keep=False)]
print(duplicados.sort_values(by=data.columns.tolist()))
```

```python
#Eliminar las filas duplicadas, quedandonos con la primera
data = data.drop_duplicates(keep='first')
data.shape
```

```python
#HOMOGENEIDAD DE DATOS
data['MODALIDAD'] = data['MODALIDAD'].str.strip().str.upper()
data['COMPLEJIDAD'] = data['COMPLEJIDAD'].str.strip().str.upper()
data['TIPO DE SERVICIO'] = data['TIPO DE SERVICIO'].str.strip().str.upper()
data['NOMBRE DE LA TAREA'] = data['NOMBRE DE LA TAREA'].str.strip().str.upper()
```

```python
#Verificar datos unicos en cada columna
print('Tipo de Servicio' + str(data['TIPO DE SERVICIO'].unique()))
print('Modalidad' + str(data['MODALIDAD'].unique()))
print('COMPLEJIDAD' + str(data['COMPLEJIDAD'].unique()))
print('Nombre de la Tarea' + str(data['NOMBRE DE LA TAREA'].unique()))
print('AÑO: ' + str(data['AÑO'].unique()))
```

```python
#Exportar el dataframe limpio a un nuevo archivo CSV
ruta_salida = "../datasets/datosRed_limpio.csv"
data.to_csv(ruta_salida, index=False, encoding="ISO-8859-1")
```
